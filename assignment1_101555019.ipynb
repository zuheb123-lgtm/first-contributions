```python
# workout_stats is a dict[str, tuple[int, int, int]]
workout_stats = {
    "Alex": (30, 45, 20),
    "Jamie": (40, 35, 30),
    "Taylor": (25, 50, 35),
} 
```

```python
for friend, minutes in list(workout_stats.items()):
    workout_stats[f"{friend}_Total"] = sum(minutes)
```

```python
friends = [name for name in workout_stats if not name.endswith("_Total")]
workout_list = [list(workout_stats[name]) for name in friends]
```

```python
yoga_running = [row[:2] for row in workout_list]
print(yoga_running)
```

```python
last_two_weightlifting = [row[2] for row in workout_list[-2:]]
print(last_two_weightlifting)
```

```python
for friend in friends:
    total = workout_stats[f"{friend}_Total"]
    if total >= 120:
        print(f'Great job staying active, {friend}!')
```

```python
name = input("Enter a friend's name: ").strip()

if name in workout_stats and not name.endswith("_Total"):
    y, r, w = workout_stats[name]
    total = workout_stats.get(f"{name}_Total", y+r+w)

    print(f"Records for {name}:")
    print(f"  Yoga: {y} minutes")
    print(f"  Running: {r} minutes")
    print(f"  Weightlifting: {w} minutes")
    print(f"  Total: {total} minutes")
else:
    print(f"Friend {name} not found in the records.")
```

```python
totals = {f: workout_stats[f"{f}_Total"] for f in friends}
highest = max(totals, key=totals.get)
lowest = min(totals, key=totals.get)

print("Highest total:", highest, totals[highest])
print("Lowest total:", lowest, totals[lowest])
```
